```python
# Useful imports and utility functions
import pychor
import galois
GF = galois.GF(2**31-1)

@pychor.local_function
def share(secret, n):
    shares = [GF.Random() for _ in range(n-1)]
    shares.append(GF(secret) - GF(shares).sum())
    return shares
```
