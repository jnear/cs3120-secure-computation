## The Petersen Graph

[The Petersen Graph](https://en.wikipedia.org/wiki/Petersen_graph) has 10 vertices and 15 edges, and can be colored with 3 colors.

![graph image](https://upload.wikimedia.org/wikipedia/commons/thumb/9/90/Petersen_graph_3-coloring.svg/500px-Petersen_graph_3-coloring.svg.png)

```python
def get_coloring(node):
    coloring = {
        # outer five nodes, clockwise from top
        0: 'red',
        1: 'blue', 
        2: 'green',
        3: 'red',
        4: 'blue',
        # inner five nodes, clockwise from top
        5: 'blue',
        6: 'red',
        7: 'red',
        8: 'green',
        9: 'green'
    }
    # only the prover knows the coloring
    return prover.constant(coloring[node])
```

## Protocol

The Prover and Verifier perform the following steps $n^2$ times, where $n$ is the number of vertices in the graph:

- **Step 1: commit.** The Prover randomizes the colors and commits to the coloring. The Prover sends the commitment to the Verifier.
- **Step 2: challenge.** The Verifier picks a random edge in the graph.
- **Step 3: response.** The Prover opens the commitment for the two vertices connected by the chosen edge. 
- **Step 4: verify.** If the two colors are the same, the Verifier rejects.
