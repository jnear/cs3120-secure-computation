```python
import pychor
import galois

GF = galois.GF((2**31)-1)
p1 = pychor.Party('p1')
p2 = pychor.Party('p2')
```

## Question 1

Describe the strategy for building a proof that `protocol_sum` securely realizes `functionality_sum` in the presence of a semi-honest adversary corrupting P1. Reference [Chapter 3 of Programming MPC](https://jnear.github.io/programming-mpc/chapters/chapter03.html).

```python
@pychor.local_function
def share(x):
    s1 = GF.Random()
    s2 = GF(x) - s1
    return s1, s2

def functionality_sum(x, y):
    Fsum = pychor.Party('Fsum')
    x.send(p1, Fsum)
    y.send(p2, Fsum)
    total = x + y
    total.send(Fsum, p1)
    total.send(Fsum, p2)
    return total

def protocol_sum(x, y):
    x1, x2 = share(x).untup(2)
    y1, y2 = share(y).untup(2)
    x2.send(p1, p2)
    y1.send(p2, p1)
    sum1 = x1 + y1
    sum2 = x2 + y2
    sum1.send(p1, p2)
    sum2.send(p2, p1)
    total = sum1 + sum2
    return total

with pychor.LocalBackend():
    x = p1.constant(10)
    y = p2.constant(15)
    result_protocol = protocol_sum(x, y)
    result_functionality = functionality_sum(x, y)
    print('Protocol Result:', result_protocol)
    print('Functionality Result:', result_functionality)
```
