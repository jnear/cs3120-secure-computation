## Question 2

Is the following protocol for 2-party summation secure? Why or why not?

```python
@pychor.local_function
def share(x):
    s1 = GF.Random()
    s2 = GF(x) - s1
    return s1, s2

def protocol_sum(x, y):
    x1, x2 = share(x).untup(2)
    y1, y2 = share(y).untup(2)
    x2.send(p1, p2)
    y1.send(p2, p1)
    sum1 = x1 + y1
    sum2 = x2 + y2
    sum1.send(p1, p2)
    sum2.send(p2, p1)
    total = sum1 + sum2
    return total

with pychor.LocalBackend():
    x = p1.constant(10)
    y = p2.constant(15)
    result = protocol_sum(x, y)
    print('Result:', result)
```
